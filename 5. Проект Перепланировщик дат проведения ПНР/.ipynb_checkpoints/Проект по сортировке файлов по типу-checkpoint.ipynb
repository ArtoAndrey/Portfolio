# Проект по перепланированию дат мероприятий ПНР 

## Описание

Проект направлен на автоматизацию процесса  перепланирования дат мероприятий ПНР  группы подготовки объектов к промышленной эксплуатации. Пользователю необходимо перепланировать мероприятия по условию, чтобы в день было не более 2х мероприятий и мероприятия проводились не чаще, чем раз в два дня

## User Story

Требуется перепланировать мероприятия из списка

## Результаты предварительного анализа

У пользователя ежемесячно возникает потребность в перепланировании мероприятий по n единицам оборудования

## Задача

Написать программу сортировки и распределения дат мероприятий

## Используемый стек технологий

- Python: Основной язык программирования для разработки.
- pandas: библиотека для работы с данными
- datetime: Библиотека для работы с данными формата datetime

```python
import pandas as pd
from datetime import datetime, timedelta
```

```python
# Пути и параметры
file_path = r'C:\Users\Andrei\Desktop\plan_pnr.xlsx'
sheet_name = 'df'
output_path = r'C:\Users\martynenkoAN\Desktop\plan_pnr_replanned.xlsx'

# Читаем исходный файл
df = pd.read_excel(file_path, sheet_name=sheet_name)

# Приводим колонку с датами к типу datetime, ошибки преобразования станут NaT
df['Дата ПНР'] = pd.to_datetime(df['Дата ПНР'], dayfirst=True, errors='coerce')

# Минимальная и максимальная дата из существующих (не NaT)
min_date = df['Дата ПНР'].min()
max_date = df['Дата ПНР'].max()

print(f"Минимальная дата: {min_date.date()}")
print(f"Максимальная дата: {max_date.date()}")

# Параметры перепланирования
max_per_day = 2
interval_days = 2

# Разделим мероприятия на две группы:
# с датой и без даты
with_date = df[df['Дата ПНР'].notna()].copy()
without_date = df[df['Дата ПНР'].isna()].copy()

# Сначала отфильтруем мероприятия с датой так, чтобы на дату было не более 2-х
# Для этого отсортируем по дате и перепланируем даты с учетом лимита max_per_day и интервала

# Сортируем по дате
with_date = with_date.sort_values('Дата ПНР').reset_index(drop=True)

new_dates = []
count_on_date = 0
current_date = min_date

for idx, row in with_date.iterrows():
    # Если текущая дата меньше даты мероприятия — сдвинем current_date к дате мероприятия
    if current_date < row['Дата ПНР']:
        current_date = row['Дата ПНР']

    if count_on_date < max_per_day:
        new_dates.append(current_date)
        count_on_date += 1
    else:
        # Переходим к следующей дате с интервалом 2 дня
        current_date += timedelta(days=interval_days)
        new_dates.append(current_date)
        count_on_date = 1

with_date['Перепланированная дата'] = new_dates

# Теперь для мероприятий без даты назначим даты начиная с минимальной даты,
# но с учетом уже занятых дат из with_date

# Соберем все занятые даты и сколько мероприятий на них запланировано
occupied_dates = with_date.groupby('Перепланированная дата').size().to_dict()

# Функция для поиска следующей свободной даты с учетом max_per_day и interval_days
def find_next_free_date(start_date, occupied, max_per_day, interval_days):
    date = start_date
    while True:
        count = occupied.get(date, 0)
        if count < max_per_day:
            return date
        else:
            date += timedelta(days=interval_days)

# Начинаем назначать даты для without_date
assigned_dates = []
current_assign_date = min_date

for _ in range(len(without_date)):
    free_date = find_next_free_date(current_assign_date, occupied_dates, max_per_day, interval_days)
    assigned_dates.append(free_date)
    # Обновляем занятость даты
    occupied_dates[free_date] = occupied_dates.get(free_date, 0) + 1

# Добавляем колонку с перепланированными датами для бездатых мероприятий
without_date['Перепланированная дата'] = assigned_dates

# Объединяем обратно два датафрейма
result_df = pd.concat([with_date, without_date], ignore_index=True)

# Сортируем по перепланированной дате для удобства просмотра
result_df = result_df.sort_values('Перепланированная дата').reset_index(drop=True)

# Сохраняем результат в новый Excel файл
result_df.to_excel(output_path, index=False)

print(f"Перепланирование завершено. Результат сохранен в {output_path}")
```
